# What is a Limit?

A limit describes the value that a function approaches as the input approaches a specific value.

Example:

$$
\lim_{x \to 2}(x+1)
$$

```python
def f(x):
    return x + 1

x_values = [
    1.9,
    1.99,
    1.999,
    2.001,
    2.01,
    2.1
]

for x in x_values:
    print(
        f"x={x}, f(x)={f(x)}"
    )
```

# Left-Hand Limit

Approach the value from the left side.

$$
\lim_{x \to a^-}f(x)
$$

```python
def f(x):
    return x**2

left_points = [
    2.0 - 0.1,
    2.0 - 0.01,
    2.0 - 0.001,
    2.0 - 0.0001
]

for x in left_points:
    print(
        f"x={x:.4f}, f(x)={f(x):.6f}"
    )
```

# Right-Hand Limit

Approach the value from the right side.

$$
\lim_{x \to a^+}f(x)
$$

```python
right_points = [
    2.0 + 0.1,
    2.0 + 0.01,
    2.0 + 0.001,
    2.0 + 0.0001
]

for x in right_points:
    print(
        f"x={x:.4f}, f(x)={f(x):.6f}"
    )
```

# Two-Sided Limit

If left and right limits are equal:

$$
\lim_{x\to a^-}f(x)
=
\lim_{x\to a^+}f(x)
$$

Then the limit exists.

```python
a = 2

left = f(a - 0.000001)

right = f(a + 0.000001)

print("Left :", left)
print("Right:", right)
```

# Piecewise Function

Functions can behave differently in different regions.

```python
def piecewise(x):

    if x < 0:
        return 1

    elif x > 0:
        return 2

    else:
        return "undefined"

values = [-1, -0.1, 0, 0.1, 1]

for value in values:
    print(
        value,
        piecewise(value)
    )
```

# Example Where Limit Does Not Exist

Limit does not exist.

```python
left_limit = piecewise(-0.00001)

right_limit = piecewise(0.00001)

print("Left Limit :", left_limit)

print("Right Limit:", right_limit)
```

# Indeterminate Form

Example:

$$
\frac{0}{0}
$$

appears frequently in Calculus.

```python
def numerator(x):
    return x**2 - 1

def denominator(x):
    return x - 1

x = 1

print(
    numerator(x)
)

print(
    denominator(x)
)
```

# Resolving the Indeterminate Form

Factor:

$$
x^2-1=(x-1)(x+1)
$$

Then simplify.

```python
def simplified(x):
    return x + 1

print(
    simplified(1)
)
```

# Symbolic Limits using SymPy

SymPy can compute limits exactly.

```python
x = sp.Symbol('x')

expr = x**2

limit_value = sp.limit(
    expr,
    x,
    3
)

print(limit_value)
```

# Symbolic Infinite Limit

```python
x = sp.Symbol('x')

expr = 1/x

result = sp.limit(
    expr,
    x,
    0,
    "+"
)

print(result)
```

# Limits and Derivatives

Derivative definition:

$$
f'(x)
=
\lim_{h\to0}
\frac{
f(x+h)-f(x)
}{h}
$$

```python
def f(x):
    return x**2

x = 3

h = 0.000001

derivative = (
    f(x+h)
    -
    f(x)
) / h

print(
    derivative
)
```

# Numerical Derivative

The result should approach:

$$
2x
$$

For:

$$
x=3
$$

Expected:

$$
6
$$

# Limits in Machine Learning

Machine Learning uses derivatives.

Derivatives are built using limits.

Therefore limits are the foundation of optimization.

```python
def loss(w):
    return (w - 5)**2

w = 2

h = 0.000001

gradient = (
    loss(w+h)
    -
    loss(w)
) / h

print(
    "Gradient:",
    gradient
)
```

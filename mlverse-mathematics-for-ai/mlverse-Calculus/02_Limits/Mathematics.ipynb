# Intuition of Limits

A limit describes the value a function approaches as the input approaches a point.

The input may never actually reach that point.

What matters is the behavior near the point.

# Limit Notation

The standard notation is:

$$
\lim_{x\to a}f(x)
$$

Read as:

"The limit of f(x) as x approaches a."

# Example

Consider:

$$
f(x)=x+1
$$

Find:

$$
\lim_{x\to2}(x+1)
$$

Substitute:

$$
2+1=3
$$

Therefore:

$$
\lim_{x\to2}(x+1)=3
$$

# Left-Hand Limit

Approaching from values smaller than a.

Notation:

$$
\lim_{x\to a^-}f(x)
$$

Examples:

1.9, 1.99, 1.999

approaching:

$$
2
$$

# Right-Hand Limit

Approaching from values greater than a.

Notation:

$$
\lim_{x\to a^+}f(x)
$$

Examples:

2.1, 2.01, 2.001

approaching:

$$
2
$$

# Two-Sided Limit

A limit exists if:

$$
\lim_{x\to a^-}f(x)
=
\lim_{x\to a^+}f(x)
$$

Both sides must agree.

# Example of a Limit That Exists

$$
f(x)=x^2
$$

Find:

$$
\lim_{x\to3}x^2
$$

Substitute:

$$
3^2=9
$$

Therefore:

$$
\lim_{x\to3}x^2=9
$$

# Example of a Limit That Does Not Exist

$$
f(x)=
\begin{cases}
1, & x<0\\
2, & x>0
\end{cases}
$$

At:

$$
x=0
$$

Left limit:

$$
1
$$

Right limit:

$$
2
$$

Since they differ:

The limit does not exist.

# Infinite Limits

Sometimes a function grows without bound.

Example:

$$
f(x)=\frac{1}{x}
$$

Then:

$$
\lim_{x\to0^+}\frac{1}{x}
=
+\infty
$$

# Negative Infinite Limits

For:

$$
f(x)=\frac{1}{x}
$$

Approaching from the left:

$$
\lim_{x\to0^-}\frac{1}{x}
=
-\infty
$$

# Limits at Infinity

Study behavior as x becomes very large.

Example:

$$
\lim_{x\to\infty}\frac{1}{x}
=
0
$$

The function gets closer and closer to zero.

# Horizontal Asymptotes

If:

$$
\lim_{x\to\infty}f(x)=L
$$

then:

$$
y=L
$$

is called a horizontal asymptote.

# Limit Law: Sum

$$
\lim_{x\to a}
[f(x)+g(x)]
=
\lim_{x\to a}f(x)
+
\lim_{x\to a}g(x)
$$

# Limit Law: Difference

$$
\lim_{x\to a}
[f(x)-g(x)]
=
\lim_{x\to a}f(x)
-
\lim_{x\to a}g(x)
$$

# Limit Law: Product

$$
\lim_{x\to a}
[f(x)g(x)]
=
\left(
\lim_{x\to a}f(x)
\right)
\left(
\lim_{x\to a}g(x)
\right)
$$

# Limit Law: Quotient

$$
\lim_{x\to a}
\frac{f(x)}{g(x)}
=
\frac{
\lim_{x\to a}f(x)
}{
\lim_{x\to a}g(x)
}
$$

provided:

$$
g(a)\neq0
$$

# Indeterminate Forms

Common forms:

$$
\frac{0}{0}
$$

$$
\frac{\infty}{\infty}
$$

$$
0\times\infty
$$

$$
\infty-\infty
$$

These require special techniques.

# Example of Indeterminate Form

$$
\lim_{x\to1}
\frac{x^2-1}{x-1}
$$

Direct substitution:

$$
\frac{0}{0}
$$

Indeterminate.

# Simplification

Factor numerator:

$$
x^2-1
=
(x-1)(x+1)
$$

Then:

$$
\frac{x^2-1}{x-1}
=
x+1
$$

Now:

$$
\lim_{x\to1}(x+1)=2
$$

# L'Hôpital's Rule

If:

$$
\frac{0}{0}
$$

or

$$
\frac{\infty}{\infty}
$$

occurs, then:

$$
\lim_{x\to a}
\frac{f(x)}{g(x)}
=
\lim_{x\to a}
\frac{f'(x)}{g'(x)}
$$

if conditions are satisfied.

# Derivatives from Limits

The derivative is defined using limits.

$$
f'(x)
=
\lim_{h\to0}
\frac{
f(x+h)-f(x)
}{h}
$$

This formula is the bridge from Limits to Calculus.

# Limits in Machine Learning

Every derivative used in machine learning originates from a limit.

Backpropagation computes gradients.

Gradients are derivatives.

Derivatives are built from limits.

Therefore limits are the foundation of neural network learning.

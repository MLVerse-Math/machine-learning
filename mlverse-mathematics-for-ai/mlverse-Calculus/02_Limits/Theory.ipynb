# Limits Theory

## What is a Limit?

A limit describes the value a function approaches as the input approaches a particular point.

Example:

$$
\lim_{x \to 2} (x+1)
$$

As x approaches 2:

$$
x+1 \to 3
$$

Therefore:

$$
\lim_{x \to 2}(x+1)=3
$$

---

## Why Limits Matter?

Limits help us understand:

- Function behavior
- Instantaneous change
- Derivatives
- Optimization

---

## Left-Hand Limit

Approaching from the left:

$$
\lim_{x\to a^-}f(x)
$$

---

## Right-Hand Limit

Approaching from the right:

$$
\lim_{x\to a^+}f(x)
$$

---

## Continuity Connection

A function is continuous when:

$$
\lim_{x\to a}f(x)=f(a)
$$

---

## Derivative Connection

Derivatives are defined using limits.

$$
f'(x)
=
\lim_{h\to0}
\frac{f(x+h)-f(x)}{h}
$$

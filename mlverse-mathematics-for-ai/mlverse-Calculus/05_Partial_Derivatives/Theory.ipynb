# Partial Derivatives Theory

## Multivariable Functions

Unlike ordinary functions:

$$
f(x)
$$

Multivariable functions depend on several variables:

$$
f(x,y)
$$

Example:

$$
f(x,y)=x^2+y^2
$$

---

## Partial Derivative

Partial derivatives measure the rate of change with respect to one variable while keeping others constant.

For x:

$$
\frac{\partial f}{\partial x}
$$

For y:

$$
\frac{\partial f}{\partial y}
$$

---

## Example

Given:

$$
f(x,y)=x^2+y^2
$$

Partial derivative with respect to x:

$$
\frac{\partial f}{\partial x}=2x
$$

Partial derivative with respect to y:

$$
\frac{\partial f}{\partial y}=2y
$$

---

## Gradient

Combining all partial derivatives:

$$
\nabla f=
\left[
\frac{\partial f}{\partial x},
\frac{\partial f}{\partial y}
\right]
$$

Gradient points toward maximum increase.

---

## Importance in AI

Every neural network training step uses gradients computed from partial derivatives.

# Definition

A function is continuous at x = a if:

$$
\lim_{x\to a}f(x)=f(a)
$$

This means the function has no break at x = a.

# Three Conditions

A function is continuous if:

1.

$$
f(a)
$$

exists

2.

$$
\lim_{x\to a}f(x)
$$

exists

3.

$$
\lim_{x\to a}f(x)=f(a)
$$

# Continuous Function Example

$$
f(x)=x^2
$$

At:

$$
x=2
$$

$$
f(2)=4
$$

and

$$
\lim_{x\to2}x^2=4
$$

Therefore:

$$
f(x)
$$

is continuous at x = 2.

# Graphical Interpretation

A continuous graph can be drawn without lifting the pen.

Examples:

- Polynomial Functions
- Exponential Functions
- Sine Functions

# Discontinuity

A discontinuity occurs when:

$$
\lim_{x\to a}f(x)
\neq
f(a)
$$

or the limit does not exist.

# Removable Discontinuity

Example:

$$
f(x)=
\frac{x^2-1}{x-1}
$$

At:

$$
x=1
$$

the function is undefined.

A hole exists in the graph.

# Jump Discontinuity

Example:

$$
f(x)=
\begin{cases}
1,&x<0\\
2,&x\ge0
\end{cases}
$$

At:

$$
x=0
$$

the function jumps.

# Infinite Discontinuity

Example:

$$
f(x)=\frac{1}{x}
$$

At:

$$
x=0
$$

the function approaches infinity.

# Oscillatory Discontinuity

Example:

$$
f(x)=\sin\left(\frac1x\right)
$$

As:

$$
x\to0
$$

the function oscillates infinitely.

# Continuity on an Interval

A function is continuous on:

$$
[a,b]
$$

if it is continuous at every point in the interval.

# Intermediate Value Theorem

If a function is continuous on:

$$
[a,b]
$$

then it takes every value between:

$$
f(a)
$$

and

$$
f(b)
$$

# Example

Consider:

$$
f(x)=x^2
$$

Between:

$$
x=1
$$

and

$$
x=3
$$

The function takes all values between:

$$
1
$$

and

$$
9
$$

# Continuity and Differentiability

If:

$$
f'(x)
$$

exists

then:

$$
f(x)
$$

must be continuous.

# Counter Example

$$
f(x)=|x|
$$

is continuous everywhere.

However:

$$
f'(0)
$$

does not exist.

# Continuity in Optimization

Optimization algorithms require smooth functions.

Continuous loss functions help models learn effectively.

# Continuity in Neural Networks

Activation functions are often continuous.

Examples:

- Sigmoid
- Tanh
- Softplus

# ReLU Discussion

$$
ReLU(x)=\max(0,x)
$$

is continuous.

However its derivative changes abruptly at:

$$
x=0
$$

# Continuity in Deep Learning

Continuous functions provide:

- Stable gradients
- Better convergence
- Smoother optimization

# Key Takeaways

✅ Continuous functions have no breaks

✅ Continuity requires:

$$
\lim_{x\to a}f(x)=f(a)
$$

✅ Discontinuities can be removable, jump, infinite, or oscillatory

✅ Differentiability implies continuity

✅ Continuity is important for machine learning and optimization

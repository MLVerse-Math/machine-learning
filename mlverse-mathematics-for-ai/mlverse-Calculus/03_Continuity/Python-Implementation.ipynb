# Continuous Function

A function is continuous if:

$$
\lim_{x \to a} f(x) = f(a)
$$

Example:

$$
f(x)=x^2
$$

```python
def f(x):
    return x**2

x = 2

print("f(2) =", f(x))
```

# Numerical Continuity Check

Approach a point from both sides.

```python
a = 2

left_points = [
    a - 0.1,
    a - 0.01,
    a - 0.001
]

right_points = [
    a + 0.1,
    a + 0.01,
    a + 0.001
]

print("Left Side")

for x in left_points:
    print(
        x,
        f(x)
    )

print("\nRight Side")

for x in right_points:
    print(
        x,
        f(x)
    )
```

# Continuity Conditions

A function is continuous if:

1. $f(a)$ exists

2. $\lim_{x \to a} f(x)$ exists

3. $\lim_{x \to a} f(x)=f(a)$

```python
a = 2

fa = f(a)

left_limit = f(a - 1e-6)

right_limit = f(a + 1e-6)

print("f(a) =", fa)

print("Left Limit =", left_limit)

print("Right Limit =", right_limit)
```

# Removable Discontinuity

Consider:

$$
f(x)=
\frac{x^2-1}{x-1}
$$

At:

$$
x=1
$$

the function is undefined.

```python
def removable(x):
    return (x**2 - 1)/(x - 1)

values = [
    0.9,
    0.99,
    0.999,
    1.001,
    1.01,
    1.1
]

for x in values:
    print(
        x,
        removable(x)
    )
```

# Simplified Function

Factor:

$$
x^2-1=(x-1)(x+1)
$$

Thus:

$$
f(x)=x+1
$$

```python
def simplified(x):
    return x + 1

print(
    simplified(1)
)
```

# Jump Discontinuity

A jump occurs when:

Left Limit ≠ Right Limit

```python
def jump_function(x):

    if x < 0:
        return 1

    return 2

left_limit = jump_function(-0.0001)

right_limit = jump_function(0.0001)

print("Left Limit :", left_limit)

print("Right Limit:", right_limit)
```

# Infinite Discontinuity

Example:

$$
f(x)=\frac1x
$$

At:

$$
x=0
$$

the function becomes unbounded.

```python
def reciprocal(x):
    return 1/x

values = [
    0.1,
    0.01,
    0.001,
    0.0001
]

for x in values:
    print(
        x,
        reciprocal(x)
    )
```

# Oscillatory Discontinuity

Example:

$$
f(x)=\sin\left(\frac1x\right)
$$

```python
def oscillatory(x):
    return np.sin(1/x)

values = [
    0.1,
    0.01,
    0.001,
    0.0005,
    0.0001
]

for x in values:
    print(
        x,
        oscillatory(x)
    )
```

# Symbolic Continuity Check using SymPy

```python
x = sp.Symbol('x')

expr = x**2

limit_value = sp.limit(
    expr,
    x,
    2
)

function_value = expr.subs(
    x,
    2
)

print("Limit :", limit_value)

print("Value :", function_value)
```

# Checking Continuity Symbolically

```python
continuous = (
    limit_value
    ==
    function_value
)

print(
    "Continuous:",
    continuous
)
```

# Intermediate Value Theorem

If a function is continuous on:

$$
[a,b]
$$

it takes every value between:

$$
f(a)
$$

and

$$
f(b)
$$

```python
def f(x):
    return x**2

a = 1
b = 3

print("f(a) =", f(a))

print("f(b) =", f(b))
```

# Searching for Intermediate Values


```python
x_values = np.linspace(
    1,
    3,
    20
)

y_values = f(x_values)

print(y_values)
```

# Continuity and Differentiability

Differentiability implies continuity.

Example:

$$
f(x)=x^2
$$

is both continuous and differentiable.

```python
def derivative(x):

    h = 1e-6

    return (
        f(x+h)
        -
        f(x)
    ) / h

print(
    derivative(2)
)
```

# Continuous Activation Functions

Neural Networks often use continuous functions.

Examples:

- Sigmoid
- Tanh
- Softplus

```python
def sigmoid(x):

    return (
        1
        /
        (
            1
            +
            np.exp(-x)
        )
    )

inputs = np.array(
    [-3,-2,-1,0,1,2,3]
)

outputs = sigmoid(inputs)

print(outputs)
```

# Hyperbolic Tangent

```python
inputs = np.array(
    [-3,-2,-1,0,1,2,3]
)

outputs = np.tanh(inputs)

print(outputs)
```

# ReLU Function

$$
ReLU(x)=\max(0,x)
$$

ReLU is continuous but not differentiable at:

$$
x=0
$$

```python
def relu(x):
    return np.maximum(
        0,
        x
    )

inputs = np.array(
    [-3,-2,-1,0,1,2,3]
)

print(
    relu(inputs)
)
```

# Continuity in Machine Learning

Continuous functions help:

- Stable optimization
- Better convergence
- Reliable gradients

Most deep learning algorithms assume continuity.

```python
def loss(w):
    return (w - 5)**2

weights = np.arange(
    0,
    11
)

loss_values = loss(weights)

for w, l in zip(
    weights,
    loss_values
):
    print(
        f"Weight={w}, Loss={l}"
    )
```

# Functions

## Mathematics

---

# What is a Function?

A function maps every input to exactly one output.

$$
f:A \rightarrow B
$$

where:

- $A$ = Domain
- $B$ = Codomain

---

# Function Notation

A function is commonly written as:

$$
f(x)
$$

Example:

$$
f(x)=x^2
$$

---

# Evaluating Functions

Given:

$$
f(x)=x^2
$$

For:

$$
x=3
$$

we get:

$$
f(3)=9
$$

---

# Domain

The set of allowable inputs.

Example:

$$
f(x)=\sqrt{x}
$$

Domain:

$$
x \ge 0
$$

---

# Range

The set of possible outputs.

Example:

$$
f(x)=x^2
$$

Range:

$$
y \ge 0
$$

---

# Polynomial Functions

General form:

$$
f(x)=a_nx^n+a_{n-1}x^{n-1}+\cdots+a_0
$$

Example:

$$
f(x)=x^3+2x+1
$$

---

# Linear Functions

General form:

$$
f(x)=mx+b
$$

where:

- $m$ = Slope
- $b$ = Intercept

---

# Quadratic Functions

General form:

$$
f(x)=ax^2+bx+c
$$

Graph:

Parabola

# Exponential Functions

General form:

$$
f(x)=a^x
$$

Example:

$$
f(x)=2^x
$$

Widely used in growth modeling.

# Logarithmic Functions

General form:

$$
f(x)=\log(x)
$$

Inverse of exponential functions.

# Trigonometric Functions

Common functions:

$$
\sin(x)
$$

$$
\cos(x)
$$

$$
\tan(x)
$$

Used in signal processing and physics.

# Composite Functions

Combining functions:

$$
(f \circ g)(x)
=
f(g(x))
$$

Example:

$$
f(x)=x^2
$$

$$
g(x)=x+1
$$

Then:

$$
f(g(x))
=
(x+1)^2
$$

# Inverse Functions

An inverse function reverses a mapping.

If:

$$
y=f(x)
$$

then:

$$
x=f^{-1}(y)
$$

Example:

$$
f(x)=x^2
$$

Inverse:

$$
f^{-1}(x)=\sqrt{x}
$$

# Functions in Machine Learning

Machine Learning models learn functions:

$$
y=f(x)
$$

Examples:

- Linear Regression
- Logistic Regression
- Neural Networks

Input:

$$
x
$$

Output:

$$
y
$$

Learning means finding the best function.

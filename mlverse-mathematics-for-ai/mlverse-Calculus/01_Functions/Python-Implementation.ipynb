# Creating a Function

A function maps inputs to outputs.

Example:

$$
f(x)=x^2
$$

```python
def square(x):
    return x**2

print(square(5))
```

# Function Parameters

Functions can accept inputs called parameters.

```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

# Return Values

Functions return outputs using the return statement.

```python
def cube(x):
    return x**3

print(cube(3))
```

# Linear Function

General form:

$$
f(x)=mx+b
$$

```python
def linear_function(x, m, b):
    return m*x + b

x = 5

y = linear_function(
    x=x,
    m=2,
    b=3
)

print(y)
```

# Evaluating Multiple Values

```python
x = np.arange(-5, 6)

y = linear_function(
    x,
    m=2,
    b=3
)

print(y)
```

# Quadratic Function

$$
f(x)=ax^2+bx+c
$$

```python
def quadratic_function(
    x,
    a,
    b,
    c
):
    return a*x**2 + b*x + c

print(
    quadratic_function(
        2,
        1,
        3,
        2
    )
)
```

# Polynomial Function

```python
def polynomial_function(x):
    
    return (
        x**3
        +
        2*x**2
        +
        x
        +
        1
    )

print(
    polynomial_function(2)
)
```

# Exponential Function

$$
f(x)=e^x
$$

```python
x = np.array([1,2,3])

y = np.exp(x)

print(y)
```

# Logarithmic Function

$$
f(x)=\log(x)
$$

```python
x = np.array([1,2,5,10])

y = np.log(x)

print(y)
```

# Trigonometric Functions

Examples:

$$
\sin(x)
$$

$$
\cos(x)
$$

```python
x = np.pi / 2

print(
    "sin(x):",
    np.sin(x)
)

print(
    "cos(x):",
    np.cos(x)
)
```

# Composite Functions

If:

$$
f(x)=x^2
$$

and

$$
g(x)=x+1
$$

Then:

$$
f(g(x))
$$

```python
def f(x):
    return x**2

def g(x):
    return x + 1

result = f(g(3))

print(result)
```

# Inverse Functions

For:

$$
f(x)=x^2
$$

The inverse is:

$$
f^{-1}(x)=\sqrt{x}
$$

```python
x = 25

inverse = np.sqrt(x)

print(inverse)
```

# Domain Example

For:

$$
f(x)=\sqrt{x}
$$

Only non-negative values are valid.

```python
x = np.array([0,1,4,9,16])

print(
    np.sqrt(x)
)
```

# Range Example

For:

$$
f(x)=x^2
$$

Outputs are always:

$$
y \ge 0
$$

```python
x = np.arange(-5,6)

y = x**2

print(y)
```

# Anonymous Functions (Lambda)

Small functions can be created using lambda expressions.

```python
square = lambda x: x**2

print(
    square(6)
)
```

# Functions with Multiple Inputs

```python
def area_rectangle(
    length,
    width
):
    return length * width

print(
    area_rectangle(
        5,
        4
    )
)# Functions Returning Multiple Values
```

# Functions Returning Multiple Values

```python
def statistics(a, b):

    total = a + b

    average = total / 2

    return total, average

total, avg = statistics(10,20)

print("Total:", total)

print("Average:", avg)
```

# Function as an Argument

```python
def apply_function(
    func,
    value
):
    return func(value)

result = apply_function(
    square,
    5
)

print(result)
```

# Functions in Machine Learning

Machine Learning learns a mapping:

$$
y=f(x)
$$

where:

- $x$ = Input Features
- $y$ = Predictions

```python
def ml_model(x):

    return 2*x + 1

sample = np.array(
    [1,2,3,4]
)

predictions = ml_model(sample)

print(predictions)
```

# Activation Function Example

ReLU:

$$
ReLU(x)=max(0,x)
$$

```python
def relu(x):

    return np.maximum(
        0,
        x
    )

x = np.array(
    [-3,-1,0,2,5]
)

print(
    relu(x)
)
```

# Sigmoid Function

$$
\sigma(x)
=
\frac{1}
{1+e^{-x}}
$$

```python
def sigmoid(x):

    return (
        1
        /
        (
            1
            +
            np.exp(-x)
        )
    )

print(
    sigmoid(2)
)
```

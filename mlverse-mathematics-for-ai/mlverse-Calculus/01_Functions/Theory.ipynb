# Functions Theory

## What is a Function?

A function is a rule that maps an input to exactly one output.

Example:

$$
f(x)=x^2
$$

Input:

$$
x=3
$$

Output:

$$
f(3)=9
$$

---

## Why Functions Matter?

Functions describe relationships between variables.

Examples:

- Temperature Prediction
- Stock Prices
- Neural Networks
- Image Classification

---

## Function Notation

Functions are usually written as:

$$
f(x)
$$

where:

- $f$ = Function name
- $x$ = Input

---

## Domain

The set of all valid inputs.

Example:

$$
f(x)=\sqrt{x}
$$

Domain:

$$
x \ge 0
$$

---

## Range

The set of all possible outputs.

Example:

$$
f(x)=x^2
$$

Range:

$$
y \ge 0
$$

---

## Functions in AI

AI models learn mathematical functions:

$$
y=f(x)
$$

where:

- $x$ = Features
- $y$ = Prediction

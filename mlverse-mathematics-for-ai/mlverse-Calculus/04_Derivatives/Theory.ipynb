# Derivatives Theory

## What is a Derivative?

A derivative measures how rapidly a function changes.

For a function:

$$
f(x)
$$

the derivative is:

$$
f'(x)
$$

---

## Formal Definition

The derivative is defined using limits:

$$
f'(x)
=
\lim_{h \to 0}
\frac{f(x+h)-f(x)}{h}
$$

---

## Geometric Interpretation

The derivative represents the slope of the tangent line.

Positive Derivative:

- Function increasing

Negative Derivative:

- Function decreasing

Zero Derivative:

- Possible maximum or minimum

---

## Why Derivatives Matter?

Derivatives power:

- Optimization
- Neural Networks
- Deep Learning
- Scientific Computing

---

## Example

For:

$$
f(x)=x^2
$$

Derivative:

$$
f'(x)=2x
$$

At:

$$
x=3
$$

Slope:

$$
6
$$

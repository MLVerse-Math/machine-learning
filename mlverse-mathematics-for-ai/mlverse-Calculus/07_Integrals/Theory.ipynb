# Integrals Theory

## What is an Integral?

Integration is the reverse process of differentiation.

If:

$$
F'(x)=f(x)
$$

then:

$$
\int f(x)\,dx = F(x)+C
$$

where:

- $F(x)$ is an antiderivative
- $C$ is the constant of integration

---

## Indefinite Integral

General form:

$$
\int f(x)\,dx
$$

Example:

$$
\int x^2\,dx
=
\frac{x^3}{3}+C
$$

---

## Definite Integral

Area accumulated between two points:

$$
\int_a^b f(x)\,dx
$$

Example:

$$
\int_0^2 x^2\,dx
=
\frac{8}{3}
$$

---

## Fundamental Theorem of Calculus

If:

$$
F'(x)=f(x)
$$

then:

$$
\int_a^b f(x)\,dx
=
F(b)-F(a)
$$

---

# Chain Rule Theory

## Composite Functions

Consider:

$$
y=f(g(x))
$$

A function inside another function.

Example:

$$
y=(x^2+1)^5
$$

Here:

$$
g(x)=x^2+1
$$

and

$$
f(u)=u^5
$$

---

## Chain Rule Formula

$$
\frac{dy}{dx}
=
\frac{dy}{du}
\times
\frac{du}{dx}
$$

or

$$
\frac{d}{dx}
f(g(x))
=
f'(g(x))
\cdot
g'(x)
$$

---

## Example

Given:

$$
y=(x^2+1)^5
$$

Derivative:

$$
\frac{dy}{dx}
=
5(x^2+1)^4
\cdot
2x
$$

$$
=
10x(x^2+1)^4
$$

---

## Why It Is Important?

Every neural network consists of nested functions.

The Chain Rule enables gradient propagation across layers.
